1.write a python program to find a length of a list using len().

```python
my_list = [10,20,30,40,50,70]
length = len(my_list)
print(f"Length of the list: {length}")
```

2.write a python program to find maximum and minimum value in a list using max() and min().

```python
my_list = [10,20,30,40,50]
max_value = max(my_list)
min_value = min(my_list)
print(f"Maximum value in the list: {max_value}")
print(f"Minimum value in the list: {min_value}")
```

3.use sum() to calculate total of all numbers in a list.

```python
my_list = [10,20,30,40,50]
total=sum(my_list)
print(f"Sum of the list: {total}")
```

4.create a list of numbers ans sort them in ascending and descending order using sort().

```python
my_list = [10,70,20,40,50]
ascending_list = sorted(my_list)
descending_list = sorted(my_list, reverse=True)
print(f"List in ascending order: {ascending_list}")
print(f"List in descending order: {descending_list}")
```

6.write a program to access the first, middle and last elements of a tuple using indexing.

```python
tuple=(70, 90, 80, 40, 50)
print(f"the first element of the tuple :{tuple[0]}")
print(f"the middle element of  the tuple :{tuple[len(tuple)//2]}")
print(f"the last element of the tuple :{tuple[-1]}")
```

10.write a program to upate a value of an existing key in a dictionary.

```python
mydict = {'name': 'John', 'age': 25, 'city': 'New York'}
print(f"Dictionary: {mydict}")
print(f"Age before updating: {mydict['age']}")
mydict['age'] = 30
print(f"Dictionary after updating age: {mydict}")
print(f"Age after updating: {mydict['age']}")
```

15.write a program to check if a key exist in a dictionary using in.

```python
mydict = {'name': 'Alice', 'age': 28, 'city': 'Los Angeles'}
print(f"Dictionary: {mydict}")
if 'age' in mydict:
    print(f"'age' key exists in the dictionary.")
else:
    print(f"'age' key does not exist in the dictionary.")
```
